```python
import  numpy as np
import matplotlib.pyplot as plt

# Create a 2D coordinate grid
x = np.linspace(-10, 10, 1000)
y = np.linspace(-10, 10, 1000)
X, Y = np.meshgrid(x, y)

# Distance from the center
R = np.sqrt(X**2 + Y**2)

# -----------------------------
# 1. Black hole event horizon
# -----------------------------
black_hole_radius = 2.5

black_hole = R < black_hole_radius

# -----------------------------
# 2. Accretion disk
# -----------------------------
# Brightness decreases with distance
disk = np.exp(-((R - 4.5) / 1.5)**2)

# Make the disk asymmetric to give it a more realistic appearance
brightness = disk * (1 + 0.5 * X / 10)

# Remove the center
brightness[black_hole] = 0

# -----------------------------
# 3. Plot
# -----------------------------
plt.figure(figsize=(8, 8))

plt.imshow(
    brightness,
    extent=[-10, 10, -10, 10],
    origin="lower",
    cmap="inferno"
)

# Draw the event horizon
circle = plt.Circle(
    (0, 0),
    black_hole_radius,
    color="black"
)

plt.gca().add_patch(circle)

# Add a thin bright ring around the black hole
ring = plt.Circle(
    (0, 0),
    black_hole_radius + 0.25,
    fill=False,
    color="orange",
    linewidth=2
)

plt.gca().add_patch(ring)

# Labels
plt.title("Pictorial Representation of a Black Hole", fontsize=16)
plt.xlabel("x")
plt.ylabel("y")

plt.axis("equal")
plt.xlim(-10, 10)
plt.ylim(-10, 10)

plt.show()
```
